### **Question 40**
(Visualization) Top 5 highest rated movies aur unke 'Meta_score' vs 'IMDB_Rating' (scaled to 100) ka ek Side-by-Side Comparison Bar Chart banayein.

```python
top5 = mov.sort_values('IMDB_Rating', ascending=False).head(5)

top5['IMDB_100'] = top5['IMDB_Rating'] * 10

plot_data = pd.melt(
    top5,
    id_vars='Series_Title',
    value_vars=['Meta_score', 'IMDB_100'],
    var_name='Type',
    value_name='Score'
)

sns.barplot(data=plot_data, x='Series_Title', y='Score', hue='Type')
```
