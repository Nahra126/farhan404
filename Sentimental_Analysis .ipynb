```python
def Check_balance(account_number,pin):
    for ac in Banks_All_Accounts:
        if ac['account_number']==account_number:
            if ac['pin']==pin:
                return f"your current balance is  Rs:{ac['intial_deposit']}"
            else:
                return 'invalid pin'
    else:
        return f'{account_number}  invalid account number'            
```

```python
def Cash_Withdrawl(account_number,pin,amount):
    for ac in Banks_All_Accounts:
        if ac['account_number']==account_number:
            if ac['pin']==pin:
                if ac['intial_deposit']>=amount:
                    ac['intial_deposit']-=amount
                    return f"your Transction Rs:{amount} is successfull!"
                else:
                    return 'Insufficient Amount'
            else:
                return 'invalid pin'
    else:
        return 'invalidd account number'
                    
```

```python
def Cash_Deposit(account_number,amount):
    for ac in Banks_All_Accounts:
        if ac['account_number']==account_number:
            ac['intial_deposit']+=amount
            return f'your amount Rs:{amount} deposit succesfully'
            
    else:
        return "invalid account_number"
    
```

```python
Cash_Deposit(4592103,50000)
```

```python
def Transfer(account_number,pin,amount,beneficiary_account):
    for ac in Banks_All_Accounts:
        if ac['account_number']==account_number:
            if ac['pin']==pin:
                if ac['intial_deposit']>=amount:
                    for bac in  Banks_All_Accounts:
                        if bac['account_number']==beneficiary_account:
                            ac['intial_deposit']-=amount
                            bac['intial_deposit']+=amount
                            return f"your amount Rs:{amount}transferred successfully!"
                            
                    else:
                        return 'invalid beneficiary account'
                else:
                    return 'insifficient ammount'
            else:
                return 'invalid pin'
    else:
        return 'invaalid account number'
                
```

```python
def Closed_account(account_number,pin):
    for index, ac in enumerate(Banks_All_Accounts):
        if ac['account_number']==account_number:
            if ac['pin']==pin:
                Closed_Accounts.append(Banks_All_Accounts .pop(index))
                return f"your account is closed please collect your amount Rs{ac['intial_deposit']}"
            else:
                return 'invalid pin'
    else:
        return 'invalid account number'
                
                
```

```python
Closed_account(7103883,1062)
```

```python
def Change_pin(account_number,pin,New_pin):
    for ac in Banks_All_Accounts:
        if ac['account_number']==account_number:
            if ac['pin']==pin:
                ac['pin'] = New_pin
                return f'your pin change Successfully'
            else:
                return 'invalid pin'
    else:
        return 'invalid account number'
            
```

# Sentiment Analysis

```python
import pandas as pd
```

```python
rvw = pd.read_excel('customer_reviews.xlsx')
rvw
```

```python
import nltk
```

```python
nltk.download('vader_lexicon')
```

```python
from nltk.sentiment.vader import SentimentIntensityAnalyzer
```

```python
sia = SentimentIntensityAnalyzer()
```

```python
def calculate_sentiment(review):
    sentiment = sia.polarity_scores(review)
    return sentiment['compound']
```

```python
review = "I absolutely at this product! It's bullshit."
print(calculate_sentiment(review))
```

```python
texts = [
    "I love this product!",     # Positive
    "This product is okay.",    # Neutral
    "I hate this product."      # Negative
]
for text in texts:
     score = sia.polarity_scores(text)
     print(f"Text:{text}\nCompound score:{score['compound']}\n")
```

```python
rvw['Sentiment_Score '] = rvw['ReviewText'].apply(calculate_sentiment)
rvw
```

```python
rvw.drop('Sentimental_Score', axis=1, inplace=True)
```

```python
def categorized_sentiment(score,rating):
    if score > 0.05:
        if rating >= 4:
            return 'Positive'
        elif rating == 3:
            return 'Mixed Positive'
        else:
            return 'Mixed Negative'
    elif score -0.05:
        if rating <= 2:
            return 'Negative'
        elif rating == 3:
            return 'Mixed Negative'
        else:
            return 'Mixed Positive'
    else:
        if rating >= 4:
            return 'Positive'
        elif rating <= 2:
            return 'Negative'
        else:
            return 'Neutral'
```

```python
rvw['Sentiment_Category'] = rvw.apply(lambda row:categorized_sentiment(row['Sentiment_Score'],row['Rating']), axis=1)
rvw
```
