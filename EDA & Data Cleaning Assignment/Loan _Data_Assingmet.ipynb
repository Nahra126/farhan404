```python
Ld['Gender'].fillna(Ld['Gender'].mode()[0],inplace=True)
```

```python
Ld['Married']=Ld['Married'].fillna(Ld['Married'].mode()[0])
```

```python
Ld['Dependents']=Ld['Dependents'].fillna(Ld['Dependents'].mode()[0])
```

```python
Ld['Self_Employed']=Ld['Self_Employed'].fillna(Ld['Self_Employed'].mode()[0])
```

```python
Ld.isna().sum()
```

```python
Ld['LoanAmount']=Ld['LoanAmount'].fillna(Ld['LoanAmount'].median())
```

```python
Ld['Loan_Amount_Term']=Ld['Loan_Amount_Term'].fillna(Ld['Loan_Amount_Term'].mode()[0])
```

```python
Ld['Credit_History']=Ld['Credit_History'].fillna(Ld['Credit_History'].median())
```
