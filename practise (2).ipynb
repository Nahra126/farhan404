```python
cart = []
total = 0

while True:
    product = input('kia chaye? or q to exit')

    if product != 'q':

        for p in Store:

            if p['item'] == product:

                print('mil jayege')
                print(f"{product} {p['price']} ka ha")

                quant = int(input('kitna chaye'))

                if quant <= p['quantity']:

                    amount = p['price'] * quant
                    p['quantity'] -= quant

                    cart.append({
                        'item': product,
                        'price': amount,
                        'quantity': quant
                    })
                    break

                else:
                    print(f"{product} {quant} itne nhi ha") 
                    break
        else:
            print('ye item nhi ha')

    else:
        print('Hamare sath shopping kerne ka shukeriya')
        break

for c in cart:
    total += c['price']

print(f"""

########## invoice ##########

Items: {[i['item'] for i in cart]}
Prices: {[i['price'] for i in cart]}
Total : {total}

""")
```

```python
lst= [1,2,3,4,5,6,7,8,9]
sqr=[]
for i in lst:
    if i%2!= 0:
        sqr.append(i**2)
print(sqr)
```

```python
[i**2 for i in lst if i%2!=0]
```

```python
lst=[1,2,3,4,5,6,7,8,9]
sqr=[]
for i in lst:
    if i%2!= 0:
        sqr.append(i**2)
    else:
        sqr.append(i)
else:
    print(sqr)
```

```python
[i**2 if i%2!=0 else i for i in lst ]
```
