```python
import pandas as pd
```

```python
data = pd.read_csv('diabetes.csv')
```

```python
data.shape
```

```python
data.info()
```

```python
data.duplicated().sum()
```

```python
X = data.drop('Outcome',axis=1)
y = data.Outcome
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X,y , test_size=0.3, random_state=42)
```

```python
from sklearn.linear_model import LogisticRegression
log_reg = LogisticRegression()
log_reg.fit(X_train,y_train)
```

```python
log_reg.coef_
```

```python
y_pred_trian = log_reg.predict(X_train)
```

```python
y_pred_trian
```

```python
pd.DataFrame({'Y_Actually':y_train , 'Y_Pred':y_pred_trian})
```

```python
correct = 0 
for t,p in zip(y_train, y_pred_trian):
    if t==p:
        correct+=1
acc = correct / len(y_train)*100
acc
```

```python
y_pred_test = log_reg.predict(X_test)
```

```python
pd.DataFrame({'Y_Actually':y_test , 'Y_Pred':y_pred_test})
```

```python
correct = 0 
for t,p in zip(y_test, y_pred_test):
    if t==p:
        correct+=1
acc = correct / len(y_test)*100
acc
```

```python
from sklearn.metrics import accuracy_score,precision_score,recall_score,confusion_matrix,classification_report
```

```python
accuracy_score(y_train, y_pred_trian)
```

```python
accuracy_score(y_test, y_pred_test)
```

```python
precision_score(y_test, y_pred_test)
```

```python
recall_score(y_test, y_pred_test)
```

```python
print(classification_report(y_train,y_pred_trian))
```
